## Task 2: Generate a Visual Summary with Diffusers (5 min)

Load a Stable Diffusion pipeline on Colab's T4 GPU. Generate a visual summary from a text prompt based on the call context.

```python
# TODO: Load Stable Diffusion and generate an image
from diffusers import StableDiffusionPipeline

# Load the pipeline (use float16 for GPU efficiency)
print("🔄 Loading Stable Diffusion pipeline... (this may take a minute)")

if torch.cuda.is_available():
    pipe = StableDiffusionPipeline.from_pretrained(
        "stabilityai/stable-diffusion-2-1",
        torch_dtype=torch.float16
    ).to("cuda")
else:
    pipe = StableDiffusionPipeline.from_pretrained(
        "stabilityai/stable-diffusion-2-1"
    )

print("✅ Stable Diffusion pipeline loaded!")
```

```python
# TODO: Generate a visual summary image
# Experiment with prompt specificity and negative prompts

prompt = "Professional product catalog layout for home goods, clean design, white background, high quality photography, commercial style"
negative_prompt = "blurry, low quality, text, watermark, distorted"

print(f"📝 Prompt: {prompt}")
print(f"🚫 Negative: {negative_prompt}")
print(f"\n🎨 Generating image...")

image = pipe(
    prompt=prompt,
    negative_prompt=negative_prompt,
    num_inference_steps=30,
    guidance_scale=7.5
).images[0]

image.save("visual_summary.png")
print(f"✅ Image saved: visual_summary.png")
print(f"   Size: {image.size}")

# Display the image
image
```

```python
# TODO: Experiment with a second prompt — more specific
prompt_v2 = "Infographic showing customer satisfaction metrics for retail, bar charts, professional design, corporate blue and green color scheme"

print(f"📝 Prompt v2: {prompt_v2}")
print(f"🎨 Generating...")

image_v2 = pipe(
    prompt=prompt_v2,
    negative_prompt=negative_prompt,
    num_inference_steps=30
).images[0]

image_v2.save("visual_summary_v2.png")
print(f"✅ Image v2 saved!")
image_v2
```

```python
### LoRA Comprehension Questions

1. **What does the `r` (rank) parameter control? What happens if you increase it?**
   `r` controls the capacity of the LoRA adapter (the size of the low-rank matrices). Increasing it gives the model more capacity to learn complex domain adaptations, but it increases memory usage and reduces the efficiency benefits of LoRA.
2. **Why does LoRA only target specific modules (q_proj, v_proj)? What would happen if you targeted all layers?**
   Most of a model's knowledge is encoded in the attention mechanism's query/value projections. Targeting only these layers captures most of the fine-tuning benefit while training very few parameters. Targeting all layers would drastically increase trainable parameters, slowing down training and increasing VRAM requirements.
3. **What does `merge_and_unload()` do? Why is it useful for deployment?**
   It permanently fuses the trained LoRA adapter weights back into the base model's original weights. This is useful for deployment because it results in a standard model artifact with no inference latency overhead, eliminating the need for the PEFT library at runtime.
4. **Approximately what percentage of parameters are trained with LoRA vs full fine-tuning?**
   LoRA typically trains around 0.1% to 1% of the model's total parameters, making it extremely efficient compared to 100% in full fine-tuning.
```

| Question | Your Answer |
|----------|-------------|
| **When would you recommend LoRA fine-tuning?** | When errors are systematic (e.g., the model consistently misunderstands the 'home goods' style), you have a labeled dataset of 500+ correct examples, and you can afford the compute time to train. |
| **When would you recommend RAG?** | When errors are factual (e.g., getting the wrong dimensions or materials), the catalog data already exists in a database, and you need to ground the model in real-time, external truth without retraining. |
| **Which factors drive the decision?** | 1) The nature of the errors (style/format vs factual knowledge). 2) Availability of training data. 3) Compute budget. 4) Frequency of data updates (RAG is better for fast-changing inventory). |
| **For BrightCart's specific case, which do you recommend?** | **RAG** (Retrieval-Augmented Generation) as the first step. |
| **Why?** | A 30% error rate on e-commerce captions is often due to the model hallucinating factual details (e.g., materials, thread count) that cannot be deduced from a photo alone. By implementing multimodal RAG, BrightCart can retrieve the actual product spec sheet and feed it to the VLM alongside the image. This changes what the model *sees* and grounds its output in truth, likely solving the problem faster and cheaper than fine-tuning. |
